Selenium is a popular open-source tool for automating web browsers. It is widely used for testing web applications across different browsers and platforms. With Selenium, you can simulate user actions like clicking buttons, entering text and navigating between pages.


Note: If you haven't installed Selenium yet, refer here to set it up on your system.

Let’s explore the core components and features of Selenium:

WebDriver
Selenium WebDriver is the core component that automates web browsers. It mimics user interaction with the web page (clicks, inputs, navigation). Each browser has its own WebDriver (e.g., ChromeDriver for Chrome, GeckoDriver for Firefox).

Example: Initializing WebDriver
This code launches Firefox browser and opens Facebook mobile login page using Selenium WebDriver.


This code launches Firefox browser and opens Facebook mobile login page using Selenium WebDriver.


from selenium import webdriver
driver = webdriver.Firefox() 
driver.get("https://mbasic.facebook.com") 
You can also use:

driver = webdriver.Chrome() # Google Chrome 

Getting Page Source
You can extract the entire HTML source of a loaded web page using driver.page_source.


from selenium import webdriver
from selenium.webdriver.common.by import By

driver = webdriver.Chrome()
driver.get("https://mbasic.facebook.com")

html = driver.page_source
print(html)

driver.get(url): Opens specified URL in the browser
html = driver.page_source: Retrieves HTML source code of the loaded web page
Locating Web Elements
To interact with a web page, you must first locate the elements (buttons, text boxes, links, etc.). Selenium offers several ways to do this by:

1. ID
An element can be located using its unique id attribute, which is often most reliable method for identifying elements on a web page.


driver.find_element(By.ID, "m_login_email")
Finds an input field with the ID m_login_email.

2. Name
An element can be located using its name attribute, which is commonly used for form fields like inputs, especially when IDs are not present or reliable.


driver.find_element(By.NAME, "pass")
This line finds the password input field with name="pass".

3. Link Text
A hyperlink can be located using its exact visible text. This method is useful for finding navigation or action links where the full text is known.


driver.find_element(By.LINK_TEXT, "Forgotten password?")
This line finds a hyperlink with the exact text “Forgotten password?”.

4. Partial Link Text
A hyperlink can be located using part of its visible text. This method is helpful when only a portion of the link text is known or consistent.


driver.find_element(By.PARTIAL_LINK_TEXT, "Forgotten")
This line finds a link that contains the partial text “Forgotten”.

5. XPath
XPath is a flexible and powerful method for locating elements in a webpage's DOM, especially when attributes like ID or name are missing or dynamic.


form = driver.find_element(By.XPATH, "//form[@id='login_form']")
email = form.find_element(By.NAME, "email")
password = form.find_element(By.NAME, "pass")
This code locates login form using XPath and then finds email and password input fields inside it using their name attributes.

6. Tag Name
An element can be located using its HTML tag name. This method is useful for selecting general elements like <title>, <body>, <a>, <div>, etc.. 


title = driver.find_element(By.TAG_NAME, "title")
This code finds the <title> tag of the page.

7. CSS Selector
An element can be located using a CSS selector, which allows selection based on element type, class, ID, attributes or even hierarchical relationships. This method is especially useful for styling-based selections or when multiple classes are used.


password = driver.find_element(By.CSS_SELECTOR, "input.bl.bm.bo.bp")
This line finds an <input> element that has all the classes bl, bm, bo and bp.

8. Class Name
An element can be located using its class name. This method is useful when elements have unique or easily identifiable class names. However, it only works with a single class name.


password = driver.find_element(By.CLASS_NAME, "bp")
This line finds an element with the class name bp.

Note: If the element has multiple class names, use By.CSS_SELECTOR instead of By.CLASS_NAME.

9. Filling Input Fields
Text can be entered into input fields using send_keys() method. This method simulates keyboard input, allowing Selenium to fill out login forms, search boxes or any text input elements.


email.send_keys("youremail@example.com")
password.send_keys("yourpassword")
These lines simulate typing an email and password into respective input fields.

10. Clicking Submit Button
A form submission or button press can be automated by locating button element and calling .click() method. This simulates a real user clicking the button.


submit_button = driver.find_element(By.XPATH, "//input[@type='submit']")
submit_button.click()
This code locates the submit button using XPath and performs a click action to submit the form.

Complete Code to Demonstrate Login Into Facebook

from selenium import webdriver
from selenium.webdriver.common.by import By

driver = webdriver.Firefox()
driver.get("https://mbasic.facebook.com")

email = driver.find_element(By.NAME, "email")
password = driver.find_element(By.NAME, "pass")

email.send_keys("youremail@example.com")
password.send_keys("yourpassword")

submit_button = driver.find_element(By.NAME, "login")
submit_button.click()
